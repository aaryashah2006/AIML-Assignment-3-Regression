```python

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    log_loss, classification_report, confusion_matrix
)

filename = "diabetes_prediction_dataset(1).csv"

if not os.path.exists(filename):
    print("Upload your diabetes CSV file:")
    uploaded = files.upload()
    filename = next(iter(uploaded))

df = pd.read_csv(filename)
df.columns = df.columns.str.strip()

print("DATASET LOADED SUCCESSFULLY")
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
print("\nDataset information:")
df.info()
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nTarget distribution:")
print(df["diabetes"].value_counts())


print("\nSTATISTICAL SUMMARY")
display(df.describe(include="all").T)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="diabetes")
plt.title("Diabetes Class Distribution")
plt.xlabel("Diabetes (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="age", hue="diabetes",
             bins=30, element="step", stat="count",
             common_norm=False)
plt.title("Age Distribution by Diabetes Status")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="diabetes", y="blood_glucose_level")
plt.title("Blood Glucose Level by Diabetes Status")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="diabetes", y="HbA1c_level")
plt.title("HbA1c Level by Diabetes Status")
plt.tight_layout()
plt.show()

target = "diabetes"
df = df.drop_duplicates().reset_index(drop=True)
df = df.dropna(subset=[target])

X = df.drop(columns=[target])
y = df[target].astype(int)

if set(y.unique()) != {0, 1}:
    raise ValueError("Expected target column 'diabetes' to contain 0 and 1.")

categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=np.number
).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])

MAX_ROWS = 20000

if len(X) > MAX_ROWS:
    X, _, y, _ = train_test_split(
        X, y,
        train_size=MAX_ROWS,
        stratify=y,
        random_state=42
    )

print("\nRecords used for modelling:", len(X))
print("Class proportions:")
print(y.value_counts(normalize=True).round(4))

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

models = {
    "SVM": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(
            max_depth=12,
            random_state=42
        ))
    ]),

    "KNN": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", KNeighborsClassifier(n_neighbors=5))
    ]),

    "Naive Bayes": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", GaussianNB())
    ])
}

results = []
trained_models = {}
predictions = {}
probabilities = {}

for name, model in models.items():
    print("\nTraining", name, "...")
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)

    predictions[name] = y_pred
    probabilities[name] = y_prob
    trained_models[name] = model

    results.append({
        "Algorithm": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, pos_label=1, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, pos_label=1, zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, y_pred, pos_label=1, zero_division=0
        ),
        "Log Loss": log_loss(y_test, y_prob, labels=[0, 1])
    })

results_df = pd.DataFrame(results)

print("\nCLASSIFICATION MODEL EVALUATION RESULTS")
display(results_df.style.format({
    "Accuracy": "{:.4f}",
    "Precision": "{:.4f}",
    "Recall": "{:.4f}",
    "F1-Score": "{:.4f}",
    "Log Loss": "{:.4f}"
}))

for name in trained_models:
    print("\n", "=" * 55)
    print("CLASSIFICATION REPORT:", name)
    print("=" * 55)
    print(classification_report(
        y_test,
        predictions[name],
        labels=[0, 1],
        target_names=["No Diabetes", "Diabetes"],
        zero_division=0
    ))

metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

results_df.set_index("Algorithm")[metrics].plot(
    kind="bar", figsize=(11, 6), ylim=(0, 1.05)
)
plt.title("Comparison of Classification Models")
plt.ylabel("Score")
plt.xlabel("Classification Algorithm")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
sns.barplot(data=results_df, x="Algorithm", y="Log Loss")
plt.title("Log Loss Comparison (Lower Is Better)")
plt.xlabel("Classification Algorithm")
plt.ylabel("Log Loss")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, (name, y_pred) in zip(axes.ravel(), predictions.items()):
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=["No Diabetes", "Diabetes"],
        yticklabels=["No Diabetes", "Diabetes"],
        ax=ax
    )
    ax.set_title(name)
    ax.set_xlabel("Predicted Class")
    ax.set_ylabel("Actual Class")

plt.tight_layout()
plt.show()

print("\nBEST MODEL ACCORDING TO EACH METRIC")

for metric in metrics:
    row = results_df.loc[results_df[metric].idxmax()]
    print(f"{metric}: {row['Algorithm']} ({row[metric]:.4f})")

best_loss = results_df.loc[results_df["Log Loss"].idxmin()]
print(f"Lowest Log Loss: {best_loss['Algorithm']} "
      f"({best_loss['Log Loss']:.4f})")

print("\nCOMPARATIVE ANALYSIS")
for _, row in results_df.iterrows():
    print("\nAlgorithm:", row["Algorithm"])
    for metric in metrics + ["Log Loss"]:
        print(f"{metric}: {row[metric]:.4f}")

best_accuracy = results_df.loc[results_df["Accuracy"].idxmax()]
best_f1 = results_df.loc[results_df["F1-Score"].idxmax()]
best_recall = results_df.loc[results_df["Recall"].idxmax()]
best_precision = results_df.loc[results_df["Precision"].idxmax()]

print("\n" + "=" * 55)
print("CONCLUSION")
print("=" * 55)
print("Highest Accuracy:", best_accuracy["Algorithm"])
print("Highest Precision:", best_precision["Algorithm"])
print("Highest Recall:", best_recall["Algorithm"])
print("Highest F1-Score:", best_f1["Algorithm"])
print("Lowest Log Loss:", best_loss["Algorithm"])
print(
    "\nThe most suitable classifier depends on the evaluation metrics, "
    "class imbalance, and the cost of false-positive and false-negative "
    "predictions. These results are experimental and do not establish "
    "clinical suitability for diagnosing diabetes."
)
```
